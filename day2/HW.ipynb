```python
##27. Remove Element
class Solution(object):
    def removeElement(self, nums, val):
        k = 0 
        for i in range(len(nums)):
          if nums[i] != val:
                nums[k] = nums[i]
                k += 1
                
        return k
```

```python
## 26
class Solution(object):
    def removeDuplicates(self, nums):
     
        if not nums:
            return 0
        write_index = 1
        
        for i in range(1, len(nums)):
            if nums[i] != nums[i - 1]:
                nums[write_index] = nums[i]
                write_index += 1
                
        return write_index
```

```python
##121. Best Time to Buy and Sell Stock
class Solution(object):
def maxProfit(self, prices):
     
        min_price = float('inf')
        max_profit = 0

        for price in prices:
            if price < min_price:
                min_price = price
            else:
                profit = price - min_price
                max_profit = max(max_profit, profit)

        return max_profit
```

```python
##88
class Solution(object):
    def merge(self, nums1, m, nums2, n):
        i = m - 1
        j = n - 1
        k = m + n - 1

        while i >= 0 and j >= 0:
            if nums1[i] > nums2[j]:
                nums1[k] = nums1[i]
                i -= 1
            else:
                nums1[k] = nums2[j]
                j -= 1
            k -= 1

        while j >= 0:
            nums1[k] = nums2[j]
            j -= 1
            k -= 1
```
